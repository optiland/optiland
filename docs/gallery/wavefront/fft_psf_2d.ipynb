# FFT PSF - 2D

```python
from optiland.psf import FFTPSF
from optiland.samples.objectives import CookeTriplet
```

```python
lens = CookeTriplet()
lens.draw()
```

```python
psf = FFTPSF(lens, field=(0, 0), wavelength=0.55)
psf.view(projection="2d", num_points=128)
```

## Cross-sections and line spread function

`view_cross_section` plots 1D slices of the PSF through its peak, along x and y. The values come from `cross_section`, which returns the positions (µm) and the PSF values along one axis. Pass `log=True` to see the diffraction rings.

```python
psf.view_cross_section()
```

`line_spread_function` integrates the PSF along the other axis, so the line spread function integrates to the same total as the PSF.

```python
import matplotlib.pyplot as plt

y, lsf = psf.line_spread_function(axis="y")

plt.plot(y, lsf)
plt.xlabel("Y (µm)")
plt.ylabel("Line Spread Function")
plt.grid(True)
plt.show()
```
