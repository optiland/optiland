# Lens Information

`LensInfoViewer` prints a surface-by-surface summary of an optical system.
`optic.info()` is the convenience entry point for the same viewer. Unlike an
analysis plot, this table describes the lens prescription, not spot quality or
image-plane irradiance.

```python
from optiland.samples.objectives import CookeTriplet
from optiland.visualization import LensInfoViewer

lens = CookeTriplet()
lens.info()
```

## Read the surface table

The row index is the surface index: 0 is the object surface, 1–6 are the six
refracting surfaces of the three lens elements, and 7 is the image surface.

| Column | Interpretation |
| --- | --- |
| Type | Surface geometry; `Stop -` marks the aperture stop (surface 4 here). |
| Comment | Optional surface annotation; blank for this sample. |
| Radius | Signed radius of curvature in mm; `inf` denotes a plane. |
| Thickness | Axial distance in mm from this surface to the next one. |
| Material | Medium immediately after the surface, not before it. |
| Conic | Dimensionless conic constant; zero for this sample's spherical surfaces. |
| Semi-aperture | Surface semi-aperture in mm, not the full diameter. |

For example, surface 1 has radius 22.01359 mm and thickness 3.25896 mm;
the medium after it is SK16. Surface 3 introduces F2 and surface 5 introduces
SK16 again. The object is at infinity, so its thickness is infinite; the image
surface has no following surface, so its thickness is `nan`. Displayed numbers
are rounded and should be checked against the sample definition for full precision.

The viewer updates the paraxial model before printing. Semi-apertures reflect the
current model (or an explicit radial aperture when present); they are not a
manufacturing specification for mechanical lens diameters. For aspheric systems,
the viewer also prints a coefficient table. This Cooke triplet has no aspheric
coefficients.

## Use the viewer directly

The explicit form below prints the same information. Neither entry point takes
display options or returns a DataFrame; use whichever form fits your workflow.
For a more extensive optical report, see the separate prescription gallery example.

```python
viewer = LensInfoViewer(lens)
viewer.view()
```
